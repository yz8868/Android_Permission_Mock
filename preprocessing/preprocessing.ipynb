```python
import os
import pandas as pd
import xml.etree.ElementTree as ET
```

```python
# Function to extract uses-permission values from an XML file
def extract_permissions(file_path):
    permissions = []
    try:
        tree = ET.parse(file_path)
        root = tree.getroot()
        for elem in root.findall("uses-permission"):
            permission = elem.get("{http://schemas.android.com/apk/res/android}name")
            if permission:
                permissions.append(permission)
    except ET.ParseError:
        print(f"Error parsing {file_path}")
    return permissions
```

```python
# Function to process all XML files in a folder
def process_folder(folder_path, output_csv="permissions.csv"):
    data = []
    for file_name in os.listdir(folder_path):
        if file_name.endswith(".xml"):  # Process only XML files
            file_path = os.path.join(folder_path, file_name)
            permissions = extract_permissions(file_path)
            for permission in permissions:
                data.append({"Name": file_name.replace("_AndroidManifest.xml", ""), "Permission": permission})

    # Create DataFrame
    df = pd.DataFrame(data)

    # Save to CSV
    df.to_csv(output_csv, index=False)
```

```python
process_folder("../data/android_perm/")
```
